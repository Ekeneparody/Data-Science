```python
### Function to find outrageous numbers to look at
def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    higher_limit = Q3 + 1.5 * IQR

    return lower_limit, higher_limit
```

```python
drawprem_games_corners = Prem_games[(Prem_games['home_goal'] == 0) & (Prem_games['away_goal'] == 0)
    & (Prem_games['home_corner'] == 0) & (Prem_games['away_corner'] == 0)]
drawprem_games_corners
```

### We can agree that the the games can end as a draw but it's highly unlikely that both teams will finish a game without playing corners and having attacking stats so we can fill it by using the mean or median of the columns 

```python
premhomecornermean = Prem_games['home_corner'].mean()
premhomecornermedian = Prem_games['home_corner'].median()
premawaycornermean = Prem_games['away_corner'].mean()
premawaycornermedian = Prem_games['away_corner'].median()

print(f'The average home corners in the prem is {premhomecornermean}')
print(f'The median home corners in the prem is {premhomecornermedian}')
print(f'The average away corners in the prem is {premawaycornermean}')
print(f'The median away corners in the prem is {premawaycornermedian}')
```

```python
drawprem_games_corners['home_corner'] = drawprem_games_corners['home_corner'].replace({
    0.0 : premhomecornermedian
})
```
