```python
#People who performed poorly
for index, row in df.iterrows():
    if row['Performance_Score'] == 'Poor':
        print(f' {index} {row['Employee_ID']} {row['First_Name']} {row['Last_Name']} is performing poorly')
```

```python
poor_count = 0
Excellent_count = 0
Good_count = 0
Average_count = 0
for index, row in df.iterrows():
    if row['Performance_Score'] == 'Poor':
        poor_count += 1
    elif row['Performance_Score'] == 'Excellent':
        Excellent_count += 1
    elif row['Performance_Score'] == 'Good':
        Good_count += 1
    elif row['Performance_Score'] == 'Average':
        Average_count += 1
        
print('The number of people that performed poorly is', poor_count)
print('The number of people that performed excellently is', Excellent_count)
print('The number of people that performed good is', Good_count)
print('The number of people that performed averagely is', Average_count)

Sum_of_People = Good_count + Excellent_count + Average_count + poor_count
print('There are',Sum_of_People,'employees')


```

```python
#Outlier function
def find_outlier(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    higher_limit = Q3 + 1.5 * IQR

    return lower_limit, higher_limit
```

```python
outlier = find_outlier(df, 'Age')
print(outlier)
```

```python
#Look at the outlier
for index, row in df.iterrows():
    if (row['Age'] < lower_limit or row['Age'] > higher_limit):
        print(row)
```
