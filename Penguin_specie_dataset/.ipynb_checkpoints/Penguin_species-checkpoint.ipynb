```python
# Outlier function
def find_outlier(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    higher_limit = Q3 + 1.5 * IQR

    return lower_limit, higher_limit
```

```python
culmenLengthOutlier = find_outlier(df, 'culmen_length_mm')
print(culmenLengthOutlier)
```

```python
cllower_limit = 25.050000000000004
clhigher_limit = 62.24999999999999

for index, row in df.iterrows():
    if(row['culmen_length_mm'] < cllower_limit or row['culmen_length_mm'] > clhigher_limit):
        print(row)
```

```python
clmean = df['culmen_length_mm'].mean()
clmedian = df['culmen_length_mm'].median()
print(clmean)
print(clmedian)
```

```python
df['culmen_length_mm'] = df['culmen_length_mm'].fillna(clmedian)
```
