## Sterling NGX performance

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Sterling Bank Stock Price History.csv')
df = data.copy()
df
```

```python
df['Date'] = pd.to_datetime(df['Date'], format = '%m/%d/%Y')
```

```python
df['Date']
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['Date']
```

```python
df['Date'] = df['Date'].dt.strftime('%d/%m/%Y')
```

```python
df
```

```python
def convert_volume(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip().upper()

    if x.endswith('K'):
        return float(x[:1]) * 1_000

    elif x.endswith('M'):
        return float(x[:1]) * 1_000_000

    else:
        return float(x)
```

```python
df['Vol.']
```

```python
df['Vol.'] = df['Vol.'].apply(convert_volume)
```

```python
df['Vol.']
```

```python
minvol = df['Vol.'].min()
maxvol = df['Vol.'].max()
medianvol = df['Vol.'].median()
meanvol = df['Vol.'].mean()

print(minvol)
print(maxvol)
print('Range:', maxvol - minvol)
print(medianvol)
print(meanvol)
```

```python
df['Vol.'] = df['Vol.'].fillna(medianvol)
```

```python
datetime = pd.to_datetime(df['Date'], format = '%d/%m/%Y')
datetimemonth = datetime.dt.to_period('M')
```

```python
results = df.groupby(datetimemonth).agg({
    'Price' : ['min', 'max']
})
results
```

```python
results.plot(kind = 'bar')
plt.xlabel('Dates')
plt.ylabel('Price')
plt.title('Prices per Month')
plt.legend(title = 'Min and Max', bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```

```python
df
```

```python
resultvol = df.groupby(datetimemonth).agg({
    'Vol.' : ['min', 'max']
})
resultvol
```

```python
resultvol.plot(kind = 'bar')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.title('Volume per Month')
plt.legend(title = 'Min and Max', bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```
