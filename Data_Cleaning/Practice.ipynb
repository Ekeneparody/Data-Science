```python
#Outlier Function
def find_outlier(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lowerlimit = Q1 - 1.5 * IQR
    higherlimit = Q3 + 1.5 * IQR

    return lowerlimit, higherlimit
```

```python
#Age outlier
ageoutlier = find_outlier(df, 'Age')
print(ageoutlier)
```

```python
agelowerlimit = 18.0
agehigherlimit = 44.0

for index, row in df.iterrows():
    if (row['Age'] < agelowerlimit or row['Age'] > agehigherlimit):
        print(row)
```
