```python
salarymean = df['Salary'].mean()
salarymedian = df['Salary'].median()
print(salarymean)
print(salarymedian)
```

```python
#Outliers for salary
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3 - Q1
print(IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
higher_limit = Q3 + 1.5 * IQR

print(lower_limit, higher_limit)
```

```python
df.loc[df['Salary'] < lower_limit, 'Salary'] = df['Salary'].median()
df.loc[df['Salary'] > higher_limit, 'Salary'] = df['Salary'].median()
```

```python
df['Salary'] = df['Salary'].fillna(salarymedian)
```
