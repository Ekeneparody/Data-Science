```python
#Outliers
#creating a function to find outliers
def find_outlier (df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    higher_limit = Q3 + 1.5 * IQR

    return lower_limit, higher_limit
```

```python
print(df['Performance_Score'].median())
```

```python
outlier = find_outlier(df, 'Performance_Score')
print (outlier)
```

```python
for index, row in df.iterrows():
    if (row['Performance_Score'] < lower_limit or  row['Performance_Score'] > higher_limit):
        print(row)
```

```python
df.loc[df['Performance_Score'] > higher_limit, 'Performance_Score'] = np.nan
```

```python
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].median())
```

```python
Performance_remark = []
for score in df['Performance_Score']:
    if score < 10:
        Performance_remark.append('Very poor')
    elif score < 20:
        Performance_remark.append('Poor')
    elif score < 30:
        Performance_remark.append('Good')
    elif score < 40:
        Performance_remark.append('Very Good')
    elif score < 50:
        Performance_remark.append('Excellent')
        
df['Performance_remark'] = Performance_remark
        
```
